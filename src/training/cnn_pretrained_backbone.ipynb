```python
import subprocess, sys, json, glob, os
import pandas as pd
```

```python
OUT = "/home/admins/rebuild_workspace/models_pretrained"
CWD = "/home/admins/lip_codebase_clean/src/training"
os.makedirs(OUT, exist_ok=True)

for name in ["mobilenetv2"]:
    for seed in [11, 22, 33, 44, 55]:
        r = subprocess.run([sys.executable, "cnn_pretrained_backbone_worker.py", name, str(seed)],
                           capture_output=True, text=True, cwd=CWD)
        print(r.stdout.strip() or r.stderr.strip()[-500:])

rows = [json.load(open(f)) for f in sorted(glob.glob(f"{OUT}/result_*.json"))]
res_p = pd.DataFrame(rows)
res_p.to_csv(f"{OUT}/pretrained_backbone_log.csv", index=False)

print("\n--- summary ---")
print(res_p.groupby('config').agg(
    total=('total_params','first'), trainable=('trainable_params','first'),
    mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
    min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
    mean_train=('final_train_acc','mean')).round(4).to_string())
```

```python
import subprocess, sys, json, glob, os
import pandas as pd
```

```python
OUT = "/home/admins/rebuild_workspace/models_augmentation"
CWD = "/home/admins/lip_codebase_clean/src/training"
os.makedirs(OUT, exist_ok=True)

GROUP_NAMES = [
                # "A_original", "B_temporal", "C_photogeo", "D_volume",
                # "E_full", "F_full_temporal", "G_full_photogeo",
                "H_vol144", "I_vol192", "J_vol288"
              ]

SEEDS = [11, 22, 33, 44, 55]

for name in GROUP_NAMES:
    for seed in SEEDS:
        r = subprocess.run(
            [sys.executable, "cnn_augmentation_comparison_worker.py", name, str(seed)],
            capture_output=True, text=True, cwd=CWD)
        print(r.stdout.strip() or r.stderr.strip()[-400:])

rows = [json.load(open(f)) for f in sorted(glob.glob(f"{OUT}/result_*.json"))]
res_a = pd.DataFrame(rows)
res_a.to_csv(f"{OUT}/augmentation_comparison_log.csv", index=False)

print("\n--- eval on 20 images per class (comparable to earlier stages) ---")
print(res_a[res_a.config.isin(["A_original","B_temporal","C_photogeo","D_volume",
                               "H_vol144","I_vol192","J_vol288"])]
      .groupby('config').agg(
          train_n=('train_samples','first'), val_n=('val_samples','first'),
          mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
          min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
          mean_train=('final_train_acc','mean')).round(4).to_string())

print("\n--- eval on 25 images per class (compare only among themselves) ---")
print(res_a[res_a.config.isin(["E_full","F_full_temporal","G_full_photogeo"])]
      .groupby('config').agg(
          train_n=('train_samples','first'), val_n=('val_samples','first'),
          mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
          min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
          mean_train=('final_train_acc','mean')).round(4).to_string())
```

```python
import subprocess, sys, json, glob, os
import pandas as pd
```

```python
OUT = "/home/admins/rebuild_workspace/models_sequence"
CWD = "/home/admins/lip_codebase_clean/src/training"
os.makedirs(OUT, exist_ok=True)

for name in ["s0_cnn_lstm128"]:
    for seed in [11, 22, 33]:
        r = subprocess.run([sys.executable, "cnn_lstm_sequence_worker.py", name, str(seed)],
                           capture_output=True, text=True, cwd=CWD)
        print(r.stdout.strip() or r.stderr.strip()[-600:])

rows = [json.load(open(f)) for f in sorted(glob.glob(f"{OUT}/result_*.json"))]
res_s = pd.DataFrame(rows)
res_s.to_csv(f"{OUT}/sequence_model_log.csv", index=False)

print("\n--- summary ---")
print(res_s.groupby('config').agg(
    params=('params','first'), batch=('batch','first'),
    mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
    min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
    mean_train=('final_train_acc','mean'), secs=('seconds','mean')).round(4).to_string())
```

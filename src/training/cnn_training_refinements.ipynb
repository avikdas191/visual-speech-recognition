```python
import sys
import pandas as pd
import subprocess, json, glob, os
```

```python
OUT = "/home/admins/rebuild_workspace/models_refinements"
os.makedirs(OUT, exist_ok=True)

for name in ["r0_baseline","r1_earlystop","r2_batchnorm","r3_dropout30","r4_dropout50"]:
    for seed in [11,22,33,44,55]:
        r = subprocess.run(
            [sys.executable, "cnn_training_refinements_worker.py", name, str(seed)],
            capture_output=True, text=True, cwd="/home/admins/lip_codebase_clean/src/training")
        print(r.stdout.strip() or r.stderr.strip()[-400:])

rows = [json.load(open(f)) for f in sorted(glob.glob(f"{OUT}/result_*.json"))]
res_r = pd.DataFrame(rows)
res_r.to_csv(f"{OUT}/training_refinements_log.csv", index=False)

print("\n--- summary ---")
print(res_r.groupby('config').agg(
    mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
    min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
    mean_train=('final_train_acc','mean'), mean_epochs=('epochs_run','mean')).round(4).to_string())
```

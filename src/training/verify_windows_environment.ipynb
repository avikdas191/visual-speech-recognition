```python
import os, numpy as np, tensorflow as tf, keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
W = r"D:\rebuild_workspace"
MODEL = os.path.join(W, "model_rebuilt_data.h5")
DATA = os.path.join(W, "06_dataset_final")
REAL = os.path.join(W, "06_test_real_only")
word_classes = ['bat','cup','drop','eat','fish','hot','jump','milk','pen','red']

model = keras.models.load_model(MODEL)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
print("params:", model.count_params())

IDG = ImageDataGenerator(rescale=1/255)
test = IDG.flow_from_directory(os.path.join(DATA, "Test"), target_size=(224,224),
    classes=word_classes, batch_size=16, class_mode='categorical', shuffle=False)
test_real = IDG.flow_from_directory(REAL, target_size=(224,224),
    classes=word_classes, batch_size=16, class_mode='categorical', shuffle=False)

lf, af = model.evaluate(test, verbose=1)
lr, ar = model.evaluate(test_real, verbose=1)

print(f"\n                  WSL (GPU)    Windows (CPU)")
print(f"full test acc      0.7150       {af:.4f}")
print(f"full test loss     0.8225       {lf:.4f}")
print(f"real only acc      0.7556       {ar:.4f}")
print(f"real only loss     0.6907       {lr:.4f}")
```

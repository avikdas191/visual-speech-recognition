```python
import pandas as pd
```

```python
log = pd.read_csv("/home/admins/rebuild_workspace/logs/frame_extraction_log.csv")

print("padded videos:", (log.padded > 0).sum())
print("  padded >  5 frames:", (log.padded > 5).sum())
print("  padded > 10 frames:", (log.padded > 10).sum())
print("  padded > 15 frames:", (log.padded > 15).sum())
print()
print("worst 10:")
print(log.nlargest(10, 'padded')[['set_name','word','source_frames','padded']].to_string(index=False))
print()
print("padding by set (sets with any padding):")
by_set = log.groupby('set_name').padded.agg(['sum','max'])
print(by_set[by_set['sum'] > 0].sort_values('sum', ascending=False).head(15).to_string())
```

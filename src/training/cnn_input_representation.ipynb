```python
import subprocess, sys, json, glob, os
import pandas as pd
```

```python
OUT = "/home/admins/rebuild_workspace/models_input_representation"
os.makedirs(OUT, exist_ok=True)
CWD = "/home/admins/lip_codebase_clean/src/training"

for name in ["v0_rgb_square", "v1_rgb_aspect", "v2_gray_square"]:
    for seed in [11,22,33,44,55]:
        r = subprocess.run([sys.executable, "cnn_input_representation_worker.py", name, str(seed)],
                           capture_output=True, text=True, cwd=CWD)
        print(r.stdout.strip() or r.stderr.strip()[-400:])

rows = [json.load(open(f)) for f in sorted(glob.glob(f"{OUT}/result_*.json"))]
res_v = pd.DataFrame(rows)
res_v.to_csv(f"{OUT}/input_representation_log.csv", index=False)

print("\n--- summary ---")
print(res_v.groupby('config').agg(
    params=('params','first'),
    mean_best_val=('best_val_acc','mean'), std_best_val=('best_val_acc','std'),
    min_val=('best_val_acc','min'), max_val=('best_val_acc','max'),
    mean_train=('final_train_acc','mean')).round(4).to_string())
```
